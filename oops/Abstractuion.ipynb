# Abstraction

Abstraction is the concept of hiding the implementation details and only showing the necessary features of the object.

this helps in reducing programming complexity and efforts

Abstraction allows us to work with an object with what an object rather than how it does it

# how can we achieve abstraction in python

# 1.Abstract class-->
A class that cannot be instantiated or that means for an abstract class we can't create the object of this class .

# 2.abstract method-->
Abstract method just because it can't accessed by the object of the abstract class itself thats why we have to call this abstract method in our child class and the implementation of the abstract method is not provided inside the abstract class but must beprovided in the child class.

the abstract method is being defined by @abstractmethod
 

```python
from abc import ABC,abstractmethod

class vechile(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop (self):
        pass

class car (vechile):
    def start(self): 
        print("the car starts with an ignite key ") 

    def stop(self):
        print("the car stops with break pedals")  

class ElectricScooter(vechile):
    def start(self):
        print("electric scooter is going to start with a power button")

    def stop(self):
        print("electric scooter stops with an electric break")           
```

```python
car1=car()
car1.start()
```

```python
scooter=ElectricScooter()
scooter.stop()
```
