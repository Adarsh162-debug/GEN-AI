# magic methods

```python
## magic methods are also called as dunder methods becoause they have double underscore __ before and after name and they are special methods inpython that allows us to customize how objects  behave
```

# these methods can help us define :
 1. object Intialization(__init__)
 2. string Representation (__str__,__repr__)
 3. Arthimetic operations(__add__,__sub__,etc)

 4.comparison operation(__eq__,__lt__)

 
 5.length,indexing,iteration(len,getitem,etc)

# why to use Magic Methods

1. makes objects behave like built in types (eg,lists,integers)
2. improve readability by allowing basic operation object1 +object2

3. simplifies code by replacing verbose function call with natural syntax.

print(object)-->str --> what is returned when print(object) is used

Arithmetic operations can be done among the objects by using magic methods

```python
## __add__
class Avenger:
    def __init__(self,name,strength):
       self.name=name
       self.strength=strength

    def __add__(self,other):
        return Avenger(self.name  + " & "+other.name, self.strength+other.strength)  

    def __str__(self):
        return f"{self.name} with combined strength {self.strength}" 

ironman=Avenger('ironman',80)
thor=Avenger("thor",100)

team=ironman + thor
print(team)
```
