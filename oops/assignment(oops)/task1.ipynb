```python
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.price=price
        self.category=category

    def __str__(self):
        return f"{self.name} + {self.price}"

    def __add__(self,other):
        return Product(self.name and other.name,self.price) + (other.price,self.category and other.category)

ironman=Product('ironman',12000,'super heman')
hulk=Product('hulk',234,'avenger')

# print(ironman)
team=ironman+hulk
print(team)
 

```
