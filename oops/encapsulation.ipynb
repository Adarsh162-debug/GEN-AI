# Encapsulation

 encapsulation is a way of wrapping data (variables and methods) inside a capsule or a class as asingle unit.

 it restricts the direct access to some of the objects components which is a mean of preventing accidental inference and misuse of the data.

 # key concept of encapsulation:
 1.public members
 2.protected members
 3.private members
 4.getter and setter methods 

# public member --> 
Acessible from anywhere or Attributes are acessible outside of the class and these members are accessible by the derived class

and these members are accessible by the derived class public member attribute is directly declared

```python
class ironman:
    def __init__(self,name,power,weapon):
        self.name=name
        self.power=power
        self.weapon=weapon

    def introduction(self): 
        print(f"hello my name is {self.name} and i have {self.weapon}")  
```

# dir()-->
it takes the input as the object name and gives us this idea that which members is public or protected or private

```python
ironman1=ironman("adarsh","intelligent","gun")
ironman1.introduction()
```

```python
dir(ironman1)
```

# private members -->
A private member is not accessible outside the classs the class and not even in the derived class and in order to have the access over these private members or to bring some modification on these members we use getter and setter method.

private members are declared by using double underscore __before the name

```python
class Avenger:
    def __init__(self,name,phone_number):
        self.name=name
        self.__phone_number=phone_number   
```

```python
avenger=Avenger('captain','2345678')
```

```python
avenger.name
```

```python
avenger.phone_number
```

# getter function

## getter method -->
it will be helping us out with accessing the private data

```python
class captain:
    def __init__(self,name,address):
        self.name=name
        self.__address=address

        ## getter method
    def accessing_add(self):
        print(f"my address is {self.__address}")
```

```python
verma=captain('verma','slc')
verma.accessing_add()
```

# setter method
--> setter method is being used for ,odifying our private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound

     #getter method
    def sound_info(self):
        print(f"the animal is going to sound as:{self.__sound}")

    def sound_modify(self,new_sound):
        self.__sound=new_sound
        print(f"new sound for the animal is :{self.__sound}")        
```

```python
animal=Animal('bruno','woof woof')
animal.sound_info()
```

```python
animal.sound_modify('dig- dig')
```

# protected data-->
it is the data which is not accessible outside the class but the protected data is going to be accessible by the derived class or child class

for having the protected member you can mention your protected member as _attribute_name or by using a single underscore

```python
class GrandParent:
    def __init__(self,name,locker_key):
        self.name=name
        self._locker_key=locker_key

    # getter method
    def locker_info(self):
        print(f"the locker key info is {self._locker_key}")

class parent(GrandParent):
    def __init__(self, name, locker_key):
        super().__init__(name, locker_key)     

    def child_reveals(self):
        print(f"the locker key information is:{self._locker_key}")   
```

```python
grandpa=GrandParent('ramu','122345')
grandpa.locker_key
```

```python
grandpa.locker_info()
```

```python
parent1=parent('raja',23456)
```

```python
parent1.child_reveals()
```
