# single inheritance-->
--> the features or the attributes from one parent class will be getting inherited into 
one child class.

parent class(properties-->attributes and       functionalities)

```python
# single inheritance
class Avenger:
    def __init__(self,name,power,):
           self.name=name
           self.power=power

    def introduction(self):
          print(f"hello my name is {self.name}") 

class ironman(Avenger):#avenger is class name  
      def __init__(self,name,power,weapon):
    # super function use karege parent class ke element ko access karne ke liye
       super().__init__(name,power)
       self.weapon=weapon
      def weapon_details(self):
           print(f"i use {self.weapon}")
            
```

# multilevel inheritance
--> the properties or methods and Attributes of the grandparent class can be inherited by theparent class and then the properties of th eparent class will beinherited by the child class. in aproper heirarchy the inheritance process is going to work for us.


grandparent(properties)
|
|
parent(properties)
|
|
child






```python
class grandparent:
    def __init__(self,nationality,religion):
        self.nationality=nationality
        self.religion=religion
    def intro(self):
        print(f'my religion is{self.religion} and nationality is{self.nationality}')

class parent(grandparent):
    def __init__(self,nationality,religion,salary):
        super().__init__(nationality,religion)
        self.salary=salary
    def  salary_detail(self):
        print(f'this is my salary of my parent is {self.salary}')

class child(parent):
    def __init__(self,nationality,religion,salary,growth):
        super().__init__(nationality,religion,salary)
        self.growth=growth
    def growth(self):
        print(f"what's your  growth in the society {self.growth}")

  
```

```python
parent1=parent('indian','hindu',24000)
parent1.intro()
```

```python
raman=child('american','muslim',12000,10)
raman.intro()
```

```python
raman.salary_detail()
```

```python
parent1.salary_detail()
```

## multiple inheritance
-->in multiple inheritance ,the properties of two parent class or multiple parent class will be getting inherited into the child class

parent_1(properties)+parent_2(properties) + parent_3(properties)....
|
|
child

```python
class ironman:
    def __init__(self,name,power,suit,weapon,nationality='america'):
        self.name=name
        self.power=power
        self.weapon=weapon
        self.nationality=nationality

    def intro(self):
        print(f'my name is {self.name} and power are:{self.power}')    

class captainamerica:  
    def __init__(self,name,power,suit,weapon,nationality='america'):
         self.name=name
         self.power=power
         self.weapon=weapon
         self.nationality=nationality

    def weapon_detail(self):
         print(f' i carry the weapon as :{self.weapon}')


class hulk(ironman,captainamerica):
    def __init__(self, name, power, suit,anger_issue, weapon, nationality):
        super().__init__(name, power, suit, weapon, nationality) 
        self.anger_issue=anger_issue


    def anger_description(self):
        print(f"about my anger issue if i am having or not then the answer is:{self.anger_issue}")

        
          
        
```
