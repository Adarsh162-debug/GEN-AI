# type of constructor
1.default constructor
2. parameterized constructor

# default constructor:
it doesnt have the parametres inside it and it is only having the self keyword inside it.

it assigns a default value to all of the objects

# parameterized constructor:
it consists of parameters inside it so that every object can have a specific attribute associated to it

it allows us to pass arguments while creating an objects

```python
class dog:
    def __init__(self):
     self.name='labra'
     self.color='black'

    def info( self):
        print(f"this dog is {self.name} and the color is{self.color}")
```

```python
pappy=dog()

```

```python
pappy.info()
```

```python
## parameterized

class dogs:

    def __init__(self,breed,color,age):
        self.breed=breed
        self.color=color
        self.age=age

    def info(self):
        print(f"my dog breed is {self.breed} and color is {self.color} and age is {self.age}")  
```

```python
labra=dogs('labra','black',2)
labra.info()
```

```python
dog2=dogs('satyam','white',1)
dog2.info()
```

```python
class Avengerss:

    origin="Avenger"

    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def origin_info(cls):
        print(f"my origin is{cls.origin}")  

    def self_info(self):
        print(f"my name is{self.name} and my power is {self.power}")    
        
```

```python
Iron_man=Avengerss("iron man","genius intellect")
Iron_man.power
```

```python
Iron_man.power
```

```python
Iron_man.self_info()
```

```python
thor=Avengerss('thor','god of thunder')
```

```python
thor.origin_info()
```

```python
# static method
class Avenger1:
    @staticmethod
    def mission():
        print("all avengers must protect the earth")

    origin='avenger'
    def __init__(amit,name,power):
        amit.name=name
        amit.power=power

    @classmethod
    def  origin_info(cls):
        print(f"the origin for the avenger is: {cls.origin}")   


    def  intro(amit):
        print(f'my name is {amit.name} and power is that have {amit.power}')


           
```

```python
iron_man=Avenger1('iron_man','genius intellect')
```

```python

#instance method call
iron_man.intro()
```

```python
# class method call
iron_man.origin_info()
```

```python
 # call static method
iron_man.mission()
```
