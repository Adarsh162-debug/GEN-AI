# types of polymorphism
1.method overriding(runtime polymorphism)
2.method overloading(compile time polymorphism)
3.operator overloading()


 method overriding-->A child class provides a specific implementation of a method that has already been defined into the parent class. this allow dynamic selection at runtime.

 

```python
class Grandparent: 
    def speak(self):
        print("hello my name is amit")

class parent(Grandparent):
    def speak(self):
        print('my name is ramu yadav')  

class Child(Grandparent):
    def speak(self):
        print('hello my name is junior amit junior')              
```

```python
def attribute_provide(objects):
    print(objects.speak())
```

```python
granda=Grandparent()
attribute_provide(granda)
```

```python
parent1=parent()
attribute_provide(parent1)
```

```python
child1=Child()
attribute_provide(child1)
```

```python
class shape:
    def area(self):
        return "this method returns the area of the shape"
```

```python
class Rectangle(shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height

    def area(self):
        return self.width*self.height    
```

```python
class circle(shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        3.14*self.radius*self.radius    
```

```python
shape=shape()
shape.area()
```

```python
rectangle=Rectangle(6,4)
rectangle.area()
```

```python
circle1=circle(5)
circle1.area()
```
