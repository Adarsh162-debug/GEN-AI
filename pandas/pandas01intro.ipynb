# pandas series 
is a column of data in a table like in excel or google sheets with labels called as indexes attached to each value.

```python
pd.Series([40,50,60],index=['bread','milk','butter'],name='price of grocery items')
```

series is a 1d array that can hold any type of data like integers,float, string(object),boolean, mixed data type

 data type --->Integers --> data in a series(column)-->quantity of item in the stock

 data type --> float --> data in a series(column)--> price or product ratings

 data type ---> strings --> column ---> product name

 data type --> boolean -->column -->is item in stock ?
 data type --> dates/ times --> column --> expiry date or  sale dates

mixed types --> milk,40,true

## creating a series in multiple ways
1.list
2.numpy array
3.dictionary
4.custom index
5.scalar value

```python
## series from list
country=['india','china','usa','pakistan']
pd.Series(country)
```

```python
batsman_runs=[48,64,72,98,114,183]
pd.Series(batsman_runs,dtype='int32')
```

```python
subject_names=['maths','physics','chemistry','biology','English']
subject_marks=[48,92,64,35,98]
pd.Series(data=subject_marks,index=subject_names,name='subject marks data',dtype='int32')
```

```python
value=np.arange(10,21)
pd.Series(value,dtype='int32')
```

```python
## dictionary to series
 
## dictionary is another data structure that stores the data in akey value pair and
# just because it stores the data in a key value pair so we don't have to explicitly mention the index 
```

```python
data_items={'math':78,'english':34,'science':92}
pd.Series(data_items)
```

```python
## custom index to series 
values=[1,2,3,4,5]
pd.Series(data=values,index=['a','b','c','d','e'])
```

```python
pd.Series(4,index=['a','b','c','d','e'])
```
