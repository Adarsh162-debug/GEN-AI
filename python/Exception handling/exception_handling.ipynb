## Error<--> Exception
 Exception Handling <--> Error Handling

 Exception Handling is a way in which if we are  getting the Errors in programming or if the user entered in input causes an error , then instead of Throwing  an Error we can raise an exception which can let the user know that the input is not correct or the given input can come up as an Error .
 As we don't want to stop the execution of the program, so we raise an error which will let the user know that the current input is not correct but the program execution won't stop.

exception are events that disrupt the normal flow of a program. they occurs when an error is encountered during program execution.

except---> whiiich is used to handle the error in a way such that the program execution shouldn't stop

else --> it will be consisting of the code which we want to run if the try block operation is working without throwing any error

finally--> finally block get executed once we are done with everything and even if we are getting an error or not getting an error still the code inside the finally block get executed

```python
num=int(input("enter your number"))
try:
    result=1/num
except Exception as ex:
    print(ex)
else:
    print(result)       
```
