# while loop


```python
# print 1 to 100
i=1
while i<100:
    print(i)
    i=i+1
```

```python
# print even no by using while loop
i=0
while(i<50):
    if(i%2==0):
        print(i)
    i=i+1
```

 # loop control statements 

```python
# print the no when the no is greater than 18 then break the loop

for i in range(1,50):
    print(i)
    if(i>=18):
        break
        
```

# continue


```python
# print even no and skip odd by continue statements
for i in range(50):
    if i%2!=0:
        continue
    else:
        print(i)
   
```

```python
# print the number from 0 to 20 but once 17 is occuring skip it

for i in range(21):
   if i==17:
        continue
   else:
     print(i)
```

### pass --> pass statement is being used for doing nothing

```python

# print the number from 0 to 20 but once  17 is occuring  do not

for i in range(21):
    if i==17:
        print("our proggram will do nothing since the value is 17 ")
        pass
    else:
        print(i)     
```

# nested loops
nested loops---> it is going to have a loop inside a loop

```python
for i  in range(4):
    for j in range(5):
        print(i,j)

```

```python
# print sum of all n natural number
sum=0
for i in range(7):
    sum=sum+i
print(f"the sum of n number is equal to:{ sum}")
```
