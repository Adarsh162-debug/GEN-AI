# nested dictionary

```python
nested_dictionary={"student1":{"name":"ram","age":21,"phone":123},
"stud2":{"name":"raj","age":23,"phone":123}}
```

```python
nested_dictionary
```

```python
for outer_key,outer_value in nested_dictionary.items():
    print(f"outer key is:{outer_key} and outer value is :{outer_value}")
```

# dictionary comprehension

```python
comprehension={x**2 for x in range(11)}
```
