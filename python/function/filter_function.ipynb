# filter
filter function is going to retieve only those which are meeting a certain condition in an iterable

 the filter function constructs an iertaor from elements of an iterable for which a condition remain true or for which function returns true value

 # syntax for filter function
 # list(filter(func,iterable))

```python
def even_function(value):
    if(value%2==0):
        return True
```

```python
iterable_list=[1,4,5,3,54,67,6,55,3,23,4,65,64,73,8,71,61,5,43,2,21,45,62,7,8]
print(list(filter(even_function,iterable_list)))
```

```python
## solving the similar problem statements with a lambda function

print(list(filter(lambda x:x%2==0,iterable_list)))
```

```python
# i want to retrive the value s which are greater then 5

print(list(filter(lambda x:x>5,iterable_list)))
```

```python
# i want to retriver all those values which are greater than 5 as well and are even as well
list_item=[1,3,4,5,6,78,87,98,12]
print(list(filter(lambda x:x>5 and x%2==0,list_item)))
```

```python
## iterating over the dictionary items
 
employees=[{'name':'amit','age':23,'city':'bhopal'},
           {'name':'anu','age':33,'city':'bhojpur'},
           {'name':'aryan','age':31,'city':'ara'}]
```

```python
def age_greater(employee):
    if employee['age']>25:
        return True
```

```python
print(list(filter(age_greater,employees)))
```
