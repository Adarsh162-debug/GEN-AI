```python
# to create function celsius to farenheit

def ftoc(temp,unit):
    
    if unit =='c':
        temperature =temp*9/5+32
        print(f"{temp} equal to ",str(temperature)+"f")
    elif unit =='f':
        temperature=round((temp-32)*5/9,3)
        print(f"{temp} equal to ",str(temperature)+"c")
    else:
        print ("not valid condition")



```

```python

print(ftoc(342,'c'))
```

```python
print(ftoc(1,'c'))
```

```python
print(ftoc(1,'f'))
```

```python
# palindrome 
def check_palindrome(string):

  string_value=string.lower().replace("","")
  # reverse eual to string value then palindrome
  if string_value==string_value[::-1]:
    print("the current string is a palindrome")
  else:
    print("the current string is not a palindrome")

```

```python
check_palindrome('aba')
```

```python
check_palindrome("assab")
```
