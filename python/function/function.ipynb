recursive function

by a  recursive function we can say that a function is going to call itself again and again

```python
# factorial of a number
def fact(num):
    fact=1;    
    if num==0 or num==1:
        print(f'the factorial of {num} is :1')
    else:
        for i in range(1,num+1):
            fact=fact*i

    return fact
    


       
```

```python
print(fact(10))
```

```python
print(fact(1))
```

```python
# written through recurssion
def factorial_recurssion(num):
    if num==1 or num==0:
        return 1
    else:
        return num*factorial_recurssion(num-1)
```

```python
number=int(input("enter your number"))
factorial_recurssion(number)
```

# function with variable length arguments

```python
def addition_positional(*num):
    add=0
    for i in num:
        add+=i
    return add
```

```python
print(addition_positional(2,3,4,5,4556677,8765))
```

```python
def  data(**user_data):
    for key, value in user_data.items():
        print(f'{key}:{value}')
```

```python
data(name='adarsh',age=12,roll_no=14)
```
