# lambda function

### it is a one liner function defined by the lambda keyword 

# syntax 


```python
lambda_function=lambda X:X**2
```

```python
type(lambda_function)
```

```python
lambda_function(5)
```

```python
sum=lambda x,y:x+y
print(sum(5,7))
```

```python
product=lambda a,x,y,z:a*x*y*z
print(product(34,67,5,9))
```
