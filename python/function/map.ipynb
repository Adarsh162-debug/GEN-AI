# map

```python
# squaring all items of a list
# list --->[1,4,5,3,8,6,9]
# []-->[1,16,25,....,81]

def square(list_item):
    empty_list=[]
    for i in list_item:
        square_item=i**2
        empty_list.append(square_item)
    print(empty_list)    

```

```python
square([1,2,4,6,8,3])
```

```python
print(list(map(lambda x:x**2,[1,3,5,7,2])))
```

```python
# map function with basic function in it
 
def square(nums):
    return nums**2

print(list(map(square,[1,3,5,8,3,7])))
```

# whenever you are going to pass the general function inside your map function as the parameter  then you dont have to use parenthesis

# while mapping the value of one iterable with another iterable we can actually apply the operation in between the values of the two iterables in a synchronous way
# [1,2,3,4] <---> [5,6,7,8]--->[6,8,10,12]

```python
number_1=[1,2,3,4]
number_2=[6,7,8,9]
print(list(map(lambda x,y:x+y,number_1,number_2)))
```

```python
# map() to convert all string values items in an iterables into an integer

map_values=(list(map(int,['1','2','3','4'])))
```

```python
print(map_values)
```

```python
employee_name=['amit','satyam','raman','hira','parth']
print(list(map(str.upper,employee_name)))
```

```python
# retrive the value inside a dictionary using map()
employees=[{'name':'amit','age':23,           'qualification':'btech'},
           {'name':'sanu','age':32,'qualification':'BBa'},
           {'name':'aryan','age':33,'qualification':'btech'}]
```

```python
def age_retiever(list_item):
    return list_item['age']
```

```python
print(list(map(age_retiever,employees)))
```
